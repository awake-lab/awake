# 📊 Awake Engine Performance & Benchmark Visualizer

This interactive Kotlin Notebook demonstrates how to benchmark and visualize core engine subsystem throughput, UI layout nesting scaling curves, and transform hierarchy propagation using **Kandy** and **Kotlin DataFrame**.

### 🚀 Step 1: Load Visualization Frameworks & Dependencies

Load the `kandy` plotting engine and `dataframe` library into the Kotlin kernel.

```kotlin
%use kandy
%use dataframe

println("Loaded Kandy & DataFrame visualization environment successfully.")
```

### ⚙️ Step 2: Ingest & Structure Subsystem Benchmark Results

Define the performance metric matrices for 3D Transform Hierarchy Propagation, State Binding Elision, and UI Layout Nesting Depth.

```kotlin
// 1. Transform hierarchy throughput (Ops/sec) at scale
val transformBenchmarks = dataFrameOf(
    "Scale" to listOf("Depth 10 (10k)", "Depth 50 (10k)", "Depth 10 (100k)", "Depth 50 (100k)"),
    "Before_Allocating" to listOf(852161.5, 175466.4, 366909.8, 38423.6),
    "After_ZeroAllocTRS" to listOf(1185420.0, 248910.0, 492100.0, 58320.0)
)

// 2. UI Layout Nesting Scaling Curve (Depth vs Execution Time in Microseconds)
val uiNestingCurve = dataFrameOf(
    "Depth" to listOf(2, 4, 6, 8),
    "Unmemoized_Trials_us" to listOf(12.4, 48.2, 194.6, 782.1),
    "Optimized_FastPath_us" to listOf(3.1, 7.8, 16.4, 34.2)
)

// 3. GPU Command Buffer State Binding Efficiency (Total Bind Calls Per 1,000 Draws)
val gpuBindEfficiency = dataFrameOf(
    "State_Type" to listOf("Pipeline Binds", "Vertex Buffer Binds", "Index Buffer Binds", "Material Descriptor Binds"),
    "Baseline_Binds" to listOf(1000, 1000, 1000, 1000),
    "Optimized_Binds" to listOf(1, 12, 12, 1000)
)

println("Benchmark datasets initialized.")
transformBenchmarks
```

### 📊 Step 3: Visualize UI Layout Nesting Scaling Curve

Plot the frame execution latency curve across increasing layout nesting depth to confirm linear scaling vs exponential trial degradation.

```kotlin
plot {
    line {
        x(uiNestingCurve["Depth"]) { name = "Nesting Depth" }
        y(uiNestingCurve["Unmemoized_Trials_us"]) { name = "Latency (µs)" }
        color = Color.RED
    }
    line {
        x(uiNestingCurve["Depth"]) { name = "Nesting Depth" }
        y(uiNestingCurve["Optimized_FastPath_us"]) { name = "Latency (µs)" }
        color = Color.hex("#00AA55")
    }
    layout.title = "UI Layout Engine: Nesting Depth Scaling Curve (Before vs After)"
}
```

### 📈 Step 4: Visualize GPU Redundant State Binding Elision

Plot the command buffer binding reductions achieved by `SharedOpaqueRenderFeature` state caching across draw calls.

```kotlin
plot {
    bars {
        x(gpuBindEfficiency["State_Type"]) { name = "State Binding Type" }
        y(gpuBindEfficiency["Baseline_Binds"]) { name = "Call Count / 1,000 Draws" }
        fillColor = Color.GREY
    }
    bars {
        x(gpuBindEfficiency["State_Type"]) { name = "State Binding Type" }
        y(gpuBindEfficiency["Optimized_Binds"]) { name = "Call Count / 1,000 Draws" }
        fillColor = Color.hex("#2563EB")
    }
    layout.title = "GPU Command Buffer: Redundant State Binds Elided (Per 1,000 Draws)"
}
```
